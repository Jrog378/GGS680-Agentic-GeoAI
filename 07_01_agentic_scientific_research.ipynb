## 1. The research loop

Based on my current thinking, this is my advice for the research cycle:

- You should pick your research area based on your interests
- Research and try to identify 1-3 research questions (RQs) you think you can answer. You need to drive this, but discussion and refinement with AI is perfectly acceptable. Just make sure you use AI to both help refine and critique/criticise your work (and to identify weakenesses early so you can adapt/improveme).
- Create a document which forms your paper structure (see `paper_template` in this repo for my .tex version). You can do this in word (or other), if you prefer. Just make sure you name your paper something like "Oughton, E.J., et al. (2026) [Paper Title]" with your name first. Do not call it "paper.tex", it needs to be understandable to other people. Do not add other people until the end as that puts you in a difficult situation if you need to remove a name. 
- Start filling out the basic document sections
- Sketch out your methodology both as a written method, and in terms of a math description, and workflow diagram. 
- Then get truth data, e.g., something you are going to want to try predict. So for land cover classification, you will need some truth data (e.g., manually labeled pixel values with text truth labels).
- Now you need to start setting up your assessment framework in order to do scientific discovery. You will need to write out decision/execution logs undertaken by the agent. You also need to instruct the agent what to do (e.g., maximize predictive accuracy, F1, mean average precision, etc.), depending on your research question. This has to be at least 1 quantitative metric that the agent can evaluate performance for, on each iteration cycle. 
- Now tell your agent(s) what you want it to do, e.g., are we going to encourage it to explore feature engineering, hyperparameter tuning, new methods, etc. or say it can do all of these?
- We expect on each iteration cycle the agent(s) to evaluate against the performance metrics we give it, critique what was carried out, either revise or stop, and also to report what happened (or at least record) with suitable evidence.
- Ideally, we might export some sort of ranking for the different things the agent(s) tried to do to maximize the performance metrics (and each entry in that ranked list would have a file somewhere providing all the detail, e.g., feature engineering techniques tried, hyperparameters tuned in specific ways, new methods tried, etc).

## 2. We don't need to re-invent the wheel

Sakana describes an agentic scientific research approach in which coding agents can investigate research directions through computational experiments, compare candidate approaches, and help create research reports. The precise implementation and capabilities are specific to the project. This is similar to what we did in previous classes, but it is **not** a general guarantee that agents autonomously produce valid or novel science.

https://pub.sakana.ai/agentic-scientist/

A useful conceptual decomposition:

| Component | Responsibility | GeoAI translation |
|---|---|---|
| Human scientist | Set research question and resources | Choose study area, scientific aims, etc. |
| Research planner / coding agent | Suggest implementable experiments | Compare spectral band baseline to index augmented model |
| Tools / compute | Run scripts, produce artifacts | rasterio, scikit-learn, Python |
| Evaluation | Quantify performance with independent evidence | Spatial holdout, per-class F1, confusion matrix |
| Memory / evidence record | Retain settings and outcomes. | Run manifest and performance tables |
| Critique and review | Detect confounds and invalid conclusions | Leakage, class imbalance, uncertain labels, etc. |

**Important:** Multiple roles or prompts do not automatically mean multiple independently running agents.

## 3. What makes an experiment scientific?

Before running code, define:
- **Question:** What specific effect are we testing?
- **Baseline:** What does the candidate need to outperform? (we generally always need this)
- **Controlled comparison:** What is held constant?
- **Data partition:** How will we avoid spatial or temporal leakage?
- **Primary metric:** What is measured, and on which withheld data?
- **Failure condition:** What result would falsify our preferred claim?
- **Stopping rule:** When will we stop experimenting?

A higher metric alone does not prove an advance, as an agent may exploit the test protocol or violate methodological constraints (e.g., make up data...).

**Discussion:** Suppose accuracy rises from 84% to 97%. What would you check before accepting the claim?

## 4. Evaluation is more than execution

| Check | Question |
|---|---|
| Software correctness | Did the code execute and produce expected shapes and units? |
| Experimental validity | Was the split independent and the comparison controlled? |
| Geographic validity | Does the result transfer to new spatial blocks or regions? |
| Scientific validity | Does the evidence support the claim? |
| Reproducibility | Can another researcher rerun the exact procedure? |

**Guardrails:** Protect raw data, e.g., you should not let the agent alter held out labels (!!! - I can't believe this is a thing). Also, it is important to require approval before changing evaluation design so the agent doesn't just fudge the performance assessment framework. A prompt instruction is much weaker than a permission or test enforced in code.
