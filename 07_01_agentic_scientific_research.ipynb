## 1. The research loop

Human-approved research question → baseline and data → propose candidate method → implement experiment → execute and log → evaluate against prespecified metrics → critique → revise or stop → report with evidence.

**Instructor prompt:** Which arrows involve model choices? Which require deterministic software? Which require a human decision?

Unlike a fixed script, a bounded agent may select a subsequent experiment after observing earlier results. The *orchestrator* enforces budgets, tool permissions, checkpoints, and stopping conditions.

## 2. Sakana AI: what is the framework?

Read the original overview: https://pub.sakana.ai/agentic-scientist/

Sakana describes an agentic scientific research approach in which coding agents can investigate research directions through computational experiments, compare candidate approaches, and help create research reports. The precise implementation and capabilities are specific to the project; it is **not** a general guarantee that agents autonomously produce valid or novel science.

A useful conceptual decomposition:

| Component | Responsibility | GeoAI translation |
|---|---|---|
| Human scientist | Set question, acceptable claims, resources | Choose study area, labels, scientific aim |
| Research planner / coding agent | Suggest implementable experiments | Compare spectral-band baseline to index-augmented model |
| Tools / compute | Run scripts, produce artifacts | rasterio, scikit-learn, Python |
| Evaluation | Quantify performance with independent evidence | Spatial holdout, per-class F1, confusion matrix |
| Memory / evidence record | Retain settings, outcomes, failed attempts | Run manifest, tables, code commit |
| Critique and review | Detect confounds and invalid conclusions | Leakage, class imbalance, uncertain labels |

**Important:** Multiple roles or prompts do not automatically mean multiple independently running agents.

## 3. What makes an experiment scientific?

Before running code, define:
- **Question:** What specific effect are we testing?
- **Baseline:** What does the candidate need to outperform?
- **Controlled comparison:** What is held constant?
- **Data partition:** How will we avoid spatial or temporal leakage?
- **Primary metric:** What is measured, and on which withheld data?
- **Failure condition:** What result would falsify our preferred claim?
- **Stopping rule:** When will we stop experimenting?

A higher metric alone does not prove an advance: an agent may exploit the test protocol or violate methodological constraints.

**Discussion:** Suppose accuracy rises from 84% to 97%. What would you check before accepting the claim?

## 4. Evaluation is more than execution

| Check | Question |
|---|---|
| Software correctness | Did the code execute and produce expected shapes and units? |
| Experimental validity | Was the split independent and the comparison controlled? |
| Geographic validity | Does the result transfer to new spatial blocks or regions? |
| Scientific validity | Does the evidence support the claim? |
| Reproducibility | Can another researcher rerun the exact procedure? |

**Guardrails:** Protect raw data; do not let the agent alter held-out labels; cap tool calls; record package versions and seeds; require approval before changing evaluation design. A prompt instruction is weaker than a permission or test enforced in code.
