# 🧠 Week 5A: From Autonomous GIS to Agentic Scientific Discovery

Agentic GeoAI is not only about automating geospatial workflows.

This notebook asks a more ambitious question:

> **What changes when an AI agent can participate in scientific reasoning — generating questions,
> proposing hypotheses, designing methods, evaluating evidence, and deciding what should be investigated next?**

The aim is **not** to remove the scientist. The aim is to examine a new division of scientific labor in
which agents perform some reasoning and execution while humans retain scientific governance, domain
judgment, and responsibility for defensible claims.

### 🎯 Learning objectives

By the end of this notebook, you should be able to:

1. distinguish **workflow autonomy** from **scientific reasoning autonomy**;
2. identify which parts of the scientific process can be delegated to agents;
3. design a multi-agent system for research-question generation and critique;
4. evaluate whether an agent-generated hypothesis is testable and scientifically meaningful;
5. design a GeoAI methodology from a research question;
6. distinguish agent self-critique from independent scientific validation;
7. identify where human oversight is most important;
8. explain how results can generate the **next research question**, rather than merely end a workflow.

⏱️ **Class format:** approximately 120 minutes, with a 2–3 minute exercise roughly every 10–15 minutes.

## 🧭 1. Two kinds of autonomy: workflow versus scientific reasoning

A workflow agent may autonomously execute a researcher-defined task:

```text
RESEARCHER DEFINES QUESTION + METHOD
                ↓
        AGENT EXECUTES GIS
                ↓
              RESULT
```

Scientific-reasoning autonomy is different:

```text
BROAD SCIENTIFIC PROBLEM
          ↓
QUESTION / HYPOTHESIS GENERATION
          ↓
CRITIQUE + PRIORITIZATION
          ↓
METHOD DESIGN
          ↓
DATA / TOOL SELECTION
          ↓
ANALYSIS / EXPERIMENT
          ↓
EVALUATION + VALIDATION
          ↓
INTERPRETATION
          ↓
NEW / REVISED QUESTION
          ↺
```

The second system is not merely doing *more GIS*. It is making decisions about **what science to do**.

## 🔬 2. Agents as participants in the scientific process

Recent agentic science systems are explicitly moving beyond productivity assistance.

Examples include systems that:
- generate hypotheses;
- search and synthesize literature;
- design experiments;
- analyze experimental data;
- update hypotheses;
- write manuscripts;
- critique research products.

A useful teaching abstraction is:

```text
OBSERVE → QUESTION → HYPOTHESIZE → DESIGN → ACT → EVALUATE → REVISE
```

Humans can enter at every stage, but the agent is no longer limited to execution.

### 📚 Research connections

- Gottweis et al. (2025), **Towards an AI co-scientist**: multi-agent generate–debate–evolve architecture for hypothesis generation.  
  https://arxiv.org/abs/2502.18864
- Ghareeb et al. (2026), **A multi-agent system for automating scientific discovery (Robin)**: integrates literature, hypothesis generation, experiment planning, analysis, and updated hypotheses.  
  https://doi.org/10.1038/s41586-026-10652-y
- Lu et al. (2026), **Towards end-to-end automation of AI research / The AI Scientist**: idea generation, code, experiments, analysis, manuscript generation, and peer review.  
  https://www.nature.com/articles/s41586-026-10265-5

## 🌍 3. Begin with a broad GeoAI problem, not a finished research question

For today's exercise, start with a broad problem such as:

- drought and inland waterway resilience;
- wildfire exposure and infrastructure;
- flood risk and transportation networks;
- remote sensing detection of critical infrastructure;
- urban heat and social vulnerability;
- land use change;
- geospatial foundation models and transferability.

The agent's first task is **not** to write prose.

It is to turn a broad problem into candidate research questions.

A useful candidate question should be:

| Criterion | Question |
|---|---|
| Importance | Would the answer matter scientifically or practically? |
| Novelty | Is there a plausible gap? |
| Testability | Could evidence contradict the hypothesis? |
| Data feasibility | Can we obtain relevant observations? |
| Spatial relevance | Is geography essential rather than decorative? |
| Tractability | Can the study be completed with available time/resources? |
| Validation | Is there a credible way to challenge the conclusion? |

## 🤖 4. Multi-agent question generation: proposer + critic

We now introduce two roles:

```text
RESEARCH QUESTION AGENT
        ↓
3–5 candidate questions
        ↓
SCIENTIFIC CRITIC
        ↓
score + challenge + request revision
        ↓
RESEARCH QUESTION AGENT
        ↓
revised question + hypothesis
```

Why not use one prompt?

Because role separation gives us:
- explicit intermediate state;
- different evaluation criteria;
- observable disagreement;
- a structured revision loop;
- an opportunity for human intervention.

But remember:

> **Two LLMs agreeing does not make a claim scientifically true.**

## ✍️ 5. Abstract generation is now a research state summary

Only after the question has survived critique do we ask a writer agent for an abstract.

The abstract should summarize the **current scientific state**, not fabricate completion.

If no results exist, the language should say:

- “we propose…”
- “we will evaluate…”
- “the study will test…”

rather than:

- “we find…”
- “results demonstrate…”

The reviewer should evaluate scientific coherence rather than grammar alone.

## 🧭 6. Methodology is where scientific reasoning becomes operational

A method agent must convert the research question into a design capable of discriminating among explanations.

For GeoAI, require the method to specify:

1. unit of analysis;
2. spatial and temporal scope;
3. outcome/target;
4. predictors/exposures;
5. data provenance and resolution;
6. preprocessing;
7. algorithm/model;
8. baseline/comparator;
9. spatial/temporal train-test design;
10. uncertainty and sensitivity;
11. external validation;
12. what result would weaken the hypothesis.

### GeoAI specific threats

- spatial autocorrelation;
- spatial leakage;
- resolution mismatch;
- CRS/unit errors;
- MAUP;
- ecological inference;
- temporal leakage;
- domain shift / lack of transferability;
- sampling bias.

## 🗺️ 7. Data and tool selection are scientific decisions

A data/tool agent should not simply produce a shopping list.

It should connect each proposed source to a methodological need:

```text
SCIENTIFIC CLAIM
      ↓
REQUIRED OBSERVATION
      ↓
DATASET
      ↓
PREPROCESSING
      ↓
METHOD / TOOL
      ↓
CHECK
```

For each dataset, ask:
- what variable does it actually measure?
- at what spatial/temporal resolution?
- with what uncertainty?
- for what geographic domain?
- is it primary evidence or validation evidence?

For each tool, ask:
- what action is permitted?
- what is the expected output?
- how can failure be detected?

## ✅ 8. Validation must constrain autonomous reasoning

As agents receive more scientific autonomy, the validation burden increases.

Use a seven layer validation stack:

| Layer | Question |
|---|---|
| Provenance | Where did the evidence come from? |
| Data QA | Are inputs plausible and complete? |
| Process verification | Did the workflow do what it claimed? |
| Spatial sanity | Does the geography make sense? |
| External validation | Does independent evidence agree? |
| Sensitivity | Does the conclusion survive reasonable changes? |
| Claim audit | Does the evidence support the language of the conclusion? |

A critic agent can help **find problems**.

It cannot make its own output independent evidence.

## 🔁 9. Scientific discovery is a loop, not a document pipeline

A mature system should be able to end an analysis with:

> “Given the evidence, what is the next scientifically valuable question?”

This is the transition from **paper-production agent** to **discovery agent**.

```text
QUESTION₀
   ↓
HYPOTHESIS₀
   ↓
METHOD₀
   ↓
EVIDENCE₀
   ↓
INTERPRETATION
   ↓
QUESTION₁
```

The new question might:
- explain an unexpected residual;
- test generalization in another region;
- distinguish two competing mechanisms;
- challenge a key assumption;
- seek stronger validation evidence.

## 👩‍🔬 10. Human governed scientific reasoning

The target is not “human removed from science.”

A useful division of labor is:

| Agent responsibility | Human responsibility |
|---|---|
| Generate candidates | Set scientific scope |
| Search/synthesize evidence | Judge relevance and credibility |
| Propose hypotheses | Decide which are worth pursuing |
| Propose methods | Review assumptions and identification |
| Execute reproducible analysis | Approve consequential actions |
| Run automated checks | Judge scientific plausibility |
| Suggest interpretation | Approve claims |
| Propose next questions | Decide research direction |

Human oversight should become **more strategic**, not merely “read every token.”

### 📚 Research connections

- Gottweis et al. (2025), AI co-scientist explicitly positions the system as augmenting scientists under scientist-provided objectives/guidance.  
  https://arxiv.org/abs/2502.18864
- Ghareeb et al. (2026), Robin is semi-autonomous and remains embedded in experimental/lab validation.  
  https://doi.org/10.1038/s41586-026-10652-y
- Oreskes et al. (1994): scientific model credibility depends on evidence and bounded claims, not proof by internal consistency.
