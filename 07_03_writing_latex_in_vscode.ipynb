# Week 7C - Writing a LaTeX paper in VS Code
**GGS 662 Agentic GeoAI | Short practical introduction | About 20-30 minutes**

**Goal:** open the supplied [paper template](paper_template/main.tex), edit it, and produce a PDF using `pdflatex` in the VS Code terminal.

LaTeX separates your document's content and structure from its typesetting. You write a plain-text `.tex` file; a compiler turns it into a `.pdf`. Your normal workflow is **edit > save > compile > inspect**.

This notebook is a reading and practical guide. Copy terminal commands into **Terminal > New Terminal** in VS Code. No Python kernel or notebook code execution is needed.

## 1. Install the editor tools and the compiler

You need two separate things:

1. **VS Code**, with the **LaTeX Workshop** extension by James Yu for editing and PDF preview.
2. A **LaTeX distribution** that supplies the actual `pdflatex` executable. On Windows, install [MiKTeX](https://miktex.org/howto/install-miktex). Reopen VS Code after installation so it sees the updated PATH. MiKTeX can prompt to install missing LaTeX packages when you first compile.

If you prefer installing into an existing Conda environment, this Windows option is available:

```bat
conda install -n hermes -c conda-forge miktex
conda activate hermes
```

Replace `hermes` with your environment name. Conda is optional; a normal MiKTeX installation works without it. The Python package named `pdflatex` is only a wrapper and does **not** install the compiler.

In the VS Code terminal, check:

```text
pdflatex --version
```

Continue once this prints compiler information. If the command is unrecognized, check the installation and PATH. In Windows Command Prompt, `where pdflatex` shows whether the executable is available.

## 3. Understand the basic document structure

A minimal LaTeX document looks like this. Read it alongside the longer `main.tex`; you do not need to replace the template with this example.

```latex
\documentclass[11pt]{article}
\usepackage[margin=1in]{geometry}

\title{My First GeoAI Paper}
\author{Student Name}
\date{October 2026}

\begin{document}
\maketitle

\section{Introduction}
This paper investigates land-cover classification.

\section{Method}
We compare predictions with independently labelled observations.

\end{document}
```

The **preamble**, before `\begin{document}`, sets the document type, packages and metadata. The **body**, inside the document environment, contains the paper. Commands begin with `\`; braces contain their arguments. A `%` starts a source comment.

In the supplied template, begin by changing `\title{...}`, `\author{...}` and `\date{...}`. Keep its package setup while you learn.

## 4. Compile and view your first PDF

Save `main.tex`. Open the terminal at the repository root and run:

```text
cd paper_template
pdflatex -interaction=nonstopmode -halt-on-error main.tex
pdflatex -interaction=nonstopmode -halt-on-error main.tex
```

If the terminal is already in `paper_template`, skip `cd paper_template`. Each successful command writes `main.pdf`. The second pass resolves cross-references that may need information from the first.

Use LaTeX Workshop's **View LaTeX PDF file** command in the Command Palette to preview the PDF, or open it in your normal PDF viewer. Check that your changed title and author appear.

These instructions compile directly with `pdflatex`. LaTeX Workshop's default Build command may use `latexmk`, which is a different tool and can require additional setup. You can use the terminal commands throughout this lesson.

## 5. Write a paragraph, a heading and an equation

Add a small practice subsection inside the document body:

```latex
\subsection{Research question}
Which method best predicts pixel land cover from satellite imagery?

We compare a baseline with an alternative method.
Our main outcome is \textbf{macro F1}; we also examine
\emph{class-specific errors}.

For illustration, a normalized difference is
\begin{equation}
I = \frac{a-b}{a+b}.
\label{eq:practice}
\end{equation}
Equation~\ref{eq:practice} defines the transformation.
```

A blank line starts a paragraph. A normal source line break behaves like a space; you do not need `\\` at the end of each line. Put inline mathematics between `$...$`, for example `$a+b$`.

Some characters have special meanings: write `\%`, `\&` and `\_` when you want a literal percent sign, ampersand or underscore in prose.

Save, compile twice, and inspect the paragraph and equation. Remove this practice material when writing your actual paper.

## 6. Use the template's citations and teaching guidance

The current `main.tex` uses `natbib` and an **inline bibliography** with `\bibitem` entries. For example:

```latex
Random Forest is an ensemble method \citep{breiman2001}.
```

The citation key must match an existing bibliography entry. With the current template, compile with `pdflatex` twice; **you do not need BibTeX**. Editing `references.bib` alone will not change its inline reference list. A later switch to a `.bib` workflow would require changing the bibliography commands as well.

The template's blue `\guide{...}` text is teaching guidance. Read it, write your own content, then remove the relevant guidance before submission. Replace the fictional results and figure placeholders with actual evidence. Check every source yourself.

## 7. Fix the first error, then compile again

| What you see | What to check |
| --- | --- |
| `pdflatex` is not recognized | Install the compiler, activate its environment if needed, and reopen VS Code |
| `File ...sty not found` | Install the named LaTeX package using MiKTeX Console or its package-install prompt |
| `Undefined control sequence` | Check the spelling of the command and whether its required package is loaded |
| `Missing $ inserted` | Check math delimiters and unescaped underscores in ordinary text |
| `Runaway argument` | Look for an unmatched `{` or `}` near the reported location |
| A citation or reference shows `?` | Check the key or label, then compile again |
| `Overfull \\hbox` | A line extends into the PDF margin; inspect the indicated text, long URL or equation |

Read the **first error** in the terminal or `main.log`; later errors may be consequences. Keep changes small and compile often. A PDF left over from an earlier successful build does not mean the current source compiled successfully.
