## 📂 1. Inspect the repository architecture

```text
week5/agents/          role specifications
week5/inputs/          human supplied problem + evidence
week5/outputs/         evolving research state
week5/src/             orchestration/state code
week5/AGENTS.md        total repository scientific rules (rules of the game)
week5/run.py           classroom entry point
```

## 🔬 2. Define your broad scientific problem

Edit `week5/inputs/problem.md`.

Do **not** begin with a fully specified method.

A good starting point is intentionally broader:

> “I want to understand whether drought creates systemic vulnerabilities in inland waterway freight networks.”

rather than:

> “Run XGBoost with these 14 variables and calculate SHAP values.”

The first leaves room for scientific reasoning.

## 🤖 3. Generate the prompt packets

Run the orchestrator.

It will **not** call a model. It creates the exact role + task + state packet that you can inspect and run in a dedicated VS Code/Codex conversation.

This makes the orchestration observable.

### Generate prompts from the terminal

First, open the terminal or command prompt in VS Code. 

With the terminal in the course repository root (the folder containing this notebook and `week5/`), run:

```
python week5/run.py
```

If your terminal is already inside `week5/`, use `python run.py` instead.
The manual runner uses only the Python standard library; no API key or extra packages are required.

We should be writing seven Markdown prompt packets under `week5/outputs/prompt_packets/`. It does not run the agents.
Paths printed by the runner are relative to `week5/`.

### Repeat this handoff for every stage

1. Save and review the previous agent's response in the output file specified below.
2. Run `python week5/run.py` again from the repository root.
3. Open the next refreshed packet and paste it into a new agent conversation.
4. Tell the agent: **Follow `week5/AGENTS.md`. Treat paths inside the packet as relative to `week5/`. Save the response to the `week5/outputs/` file specified for this stage.**

The packets are snapshots of the files available when the command runs. Later packets initially have blank sections because earlier responses do not exist yet. Refresh them after each saved response or revision, and inspect the required state before using the next packet. Regeneration overwrites prompt packets, but does not overwrite your saved research outputs.

The shared rules are not embedded in the packets. If the agent cannot read workspace files, paste the contents of `week5/AGENTS.md` alongside the packet and save the returned response yourself.

First save the preceding output and refresh the packets with `python week5/run.py` from the repository root.

`week5/outputs/prompt_packets/02_scientific_critic.md`

`week5/outputs/question_review.md`

For example, in the agent window prompt bar try use:

The critic is not a grammar checker. It should attack the scientific design.

Look for:
- unfalsifiable hypotheses;
- weak novelty claims;
- scale mismatch;
- causal overreach;
- no independent validation;
- unavailable data;
- spatial leakage risks.

## 🔁 10. Close the loop: what should the agent investigate next?

Ask the critic/validator:

> Given the proposed methodology and its largest unresolved threat to validity, propose one follow-on
> research question that would most reduce scientific uncertainty.

This is the step that turns a linear workflow into a discovery loop.
