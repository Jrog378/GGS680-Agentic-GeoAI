## 🧪 5. Run Agent 2: scientific critique

Now run:

`outputs/prompt_packets/02_scientific_critic.md`

Save the response as:

`outputs/question_review.md`

For example: **Follow week5/AGENTS.md. Save your response to week5/outputs/question_review.md.**

The critic is not a grammar checker. It should attack the scientific design.

Look for:
- unfalsifiable hypotheses;
- weak novelty claims;
- scale mismatch;
- causal overreach;
- no independent validation;
- unavailable data;
- spatial leakage risks.

### 📚 Research connections

- Shinn et al. (2023), Reflexion: feedback can guide revision.
- Roberts et al. (2017), spatial/temporal cross-validation: https://doi.org/10.1111/ecog.02881
- Oreskes et al. (1994), model validation and bounded claims: https://doi.org/10.1126/science.263.5147.641
