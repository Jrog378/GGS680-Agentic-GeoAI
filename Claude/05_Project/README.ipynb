**Form rules:**
- Keep the `### <code>. <title>` headings, because the script finds each field by its code.
- Lines starting with `> 💡` are hints and are ignored.
- A scaffold you leave unchanged counts as blank.
- **Evidence:** a line in A4 that is just a file path (e.g. `literature/Kapoor et al. 2024. AI Agents That Matter.pdf`) embeds that file. Everything in `evidence/` is embedded automatically, and PDFs are read with pypdf. Each file is capped at 20,000 characters.

## Commands and status words

```
python Claude/05_Project/run.py                  # status + rebuild packets + next step
python Claude/05_Project/run.py --redo N         # archive stage N's output and packet, rebuild it as a revision
python Claude/05_Project/run.py --accept         # mark all saved outputs up to date (after a change that doesn't matter)
python Claude/05_Project/run.py --copy           # also copy the next packet to the clipboard
python Claude/05_Project/run.py --rebuild-form   # regenerate the form layout, keeping your answers
python Claude/05_Project/run.py --form other.md --evidence-chars 50000
```

| Status | Meaning |
|---|---|
| done | Output saved; nothing it was built from has changed |
| ready | Inputs exist; packet written; run it next |
| waiting | An earlier output is missing |
| locked | Stages 4–8, until every H4 box is ticked |
| stale | An output it used changed, or the form fields it used changed (A1–A3, L1, evidence, decisions, notes). Run `--redo N` if it matters, `--accept` if not |
