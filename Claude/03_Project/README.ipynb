# 03_Project — Week 3 fiber route planning

**Task source:** the section 6 prompt in `03_02_ggs662_agentic_application_route_planning.ipynb`, following the Week 3 rules in `AGENTS.md`. Discussion questions are from 03_02 section 8.

**What we did:** compared the shortened baseline (`route1_short.gpkg`) with shortest paths along roads, power lines and rail between the same fixed endpoints. Each layer is its own graph, disconnected lines are joined by bounded straight connectors, and costs use a uniform USD 10/m rate. We then reran the analysis with only the gap tolerance changed.

## How to run
All arguments have defaults. The inputs are read from the repo root, which the script only reads from.
```
python week3_routing.py                                                    # original, gap 50 m -> outputs/default
python week3_routing.py --gap-tolerance-m 25  --out outputs/sensitivity_gap_25m
python week3_routing.py --gap-tolerance-m 250 --out outputs/gap_tolerance_250m
python plot_side_by_side.py                                                # baseline vs each route -> outputs/default/route_side_by_side.png
```
